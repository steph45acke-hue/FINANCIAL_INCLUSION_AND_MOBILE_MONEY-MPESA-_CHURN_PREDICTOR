```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Connect to our MySQL database engine
engine = create_engine("mysql+pymysql://root:stephen0111301468@localhost:3306/mpesa_churn_db")

# 2. Write the corrected RFM query using exact table columns
rfm_query = """
SELECT 
    c.customer_id,
    MAX(c.age) AS age,
    MAX(c.kyc_tier) AS kyc_tier,
    MAX(c.churn) AS churn,
    COUNT(t.transaction_id) AS frequency,
    COALESCE(SUM(t.amount_kes), 0) AS monetary_value,
    DATEDIFF('2026-09-01', MAX(t.timestamp)) AS recency_days
FROM customers c
LEFT JOIN transactions t ON c.customer_id = t.customer_id
GROUP BY c.customer_id;
"""

# 3. Execute the query and load results directly into a Pandas DataFrame
features_df = pd.read_sql(rfm_query, con=engine)

# 4. Preview the first few rows of our engineered feature table
print(features_df.head())
```
