```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Load our saved raw CSV data back into Pandas dataframes

customers_df = pd.read_csv("data/raw/customer_demographics.csv")
transactions_df = pd.read_csv("data/raw/mpesa_transactions.csv")

# 2. Establish connection to your local MySQL database

engine = create_engine("mysql+pymysql://root:stephen0111301468@localhost:3306/mpesa_churn_db")

# 3. Push the DataFrames directly into SQL tables
customers_df.to_sql("customers", con=engine, if_exists="replace", index=False)
transactions_df.to_sql("transactions", con=engine, if_exists="replace", index=False)

print("✅ Successfully ingested raw data into MySQL tables: 'customers' and 'transactions'!")
```
